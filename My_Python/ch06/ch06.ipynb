### 1. 정적 웹 페이지 크롤링

````python
# 초기 1회만
# !pip install beautifulsoup4

from bs4 import BeautifulSoup # BeautifulSoup 임포트

# 연습용 html. 문자열.
html='<h1 id="title">한빛출판네트워크</h1><div class="top"><ul class="menu"><li><a href="http://www.hanbit.co.kr/member/login.html" class="login">로그인</a></li></ul><ul class="brand"><li><a href="http://www.hanbit.co.kr/media/">한빛미디어</a></li><li><a href="http://www.hanbit.co.kr/academy/">한빛아카데미</a></li></ul></div>'
soup = BeautifulSoup(html, 'html.parser') # BeautifulSoup 객체를 생성. html.parser는 html을 파싱하는 애.

print(type(soup)) # <class 'bs4.BeautifulSoup'>
print(soup.prettify()) # 태그 시작과 끝을 맞춰서 들여쓰기
````

````python
# 1. 태그 파싱하기 - 지정된 한 개의 태그만 파싱(첫번째를 가져옴)

tag_h1 = soup.h1
print(f'<h1> 태그: {tag_h1}') # <h1> 태그만 가져옴

tag_div = soup.div
print(f'<div> 태그: {tag_div}') # <div> 태그만 가져옴

tag_ul = soup.ul
print(f'<ul> 태그: {tag_ul}') # <ul> 태그만 가져옴

tag_li = soup.li
print(f'<li> 태그: {tag_li}') # <li> 태그만 가져옴

tag_a = soup.a
print(f'<a> 태그: {tag_a}') # <a> 태그만 가져옴

print(soup.div.prettify()) # <div> 태그만 가져와 시각화
````

````python
# 2. 태그 파싱하기 - 지정된 태그를 모두 파싱

tag_ul_all = soup.find_all("ul") # <ul> 태그 다 가져와 리스트로 만듦
print(tag_ul_all, end='\n\n')

tag_li_all = soup.find_all("li") # <li> 태그 다 가져와 리스트 만듦
print(tag_li_all, end='\n\n')

tag_a_all = soup.find_all("a") # <a> 태그 다 가져와 리스트 만듦
print(tag_a_all, end='\n\n')
````

````python
# 3. 태그 파싱하기 - 속성을 이용하여 파싱

tag_a.attrs # attributes. 속성 이름과 속성값으로 딕셔너리를 구성한다
print(tag_a.attrs)

print(tag_a['href']) # hypertext reference. 하나의 고유한 주소나 식별자만 가질 수 있는 싱글 값 속성
print(tag_a['class']) # 공백을 기준으로 여러 개의 값을 가질 수 있는 다중 값 속성 -> 리스트를 반환. ex. class="login btn" -> 결과는 ['login', 'btn']

tag_ul_2 = soup.find("ul", attrs={'class': 'brand'}) # ul 중 class 속성이 brand인 것을 찾아라
print(tag_ul_2)

title = soup.find(id="title")
print(title)
print(title.string) # 태그 제외한 텍스트만 뽑음

# div로 파싱. 이후 ul 중 class 속성이 brand인 것을 파싱. 이후 li 파싱.
li_list = soup.select("div>ul.brand>li") # class 속성이면 ul.brand이고, 만일 id 속성이면 h1#title
print(li_list)

for li in li_list:
    print(li.string)


````

#### 참고) 크롤링 허용 여부 확인하기
- 크롤링 전, 크롤링 허용 여부를 확인하기 위해 주소창에 '크롤링할 주소/robots.txt'를 입력
- robots.txt가 없는 경우: 수집에 대한 정책이 없는 것으로, 크롤링 해도 됨

<table>
    <tr>
        <th>표시</th>
        <th>허용 여부</th>
    </tr>
    <tr>
        <td><i>User-agent:*</i> -> 모든 에이전트에 대해 아래 규칙 적용<br><i>Allow:/</i></td>
        <td rowspan="2">모든 접근 허용</td>
    </tr>
    <tr>
        <td><i>User-agent:*</i><br><i>Disallow: </i></td>
    </tr>
    <tr>
        <td><i>User-agent:*</i><br><i>Disallow:/</i></td>
        <td>모든 접근 금지</td>
    </tr>
    <tr>
        <td><i>User-agent:*</i><br><i>Disallow:/user/</i></td>
        <td>특정 디렉토리만 접근 금지</td>
    </tr>
    <tr>
        <td><i>User-agent:*</i><br><i>Disallow:/images</i><br><i>Allow:/images/public</i></td>
        <td>/private 이하 모든 경로를 차단하되, images/public 밑은 허용.</td>
    </tr>
</table>

- 접근 제어(Access Control) 방식
    - 화이트리스트 기반: 이것만 돼
    - 블랙리스트 기반: 이것만 안돼
- robots.txt 주요 규칙
    - 기본 구조
        - User-agent: 규칙이 적용될 크롤러. *는 모든 크롤러에 적용. ex. googlebot, bingbot
        - disallow: 접근을 차단할 경우 -> 화이트리스트 기반 동작
        - allow: 접근을 허용할 경로 -> 블랙리스트 기반 동작
            - 보통 disallow로 넓은 범위를 차단하고 특정 경로는 허용할 때 사용
    - 패턴 매칭 규칙:
        - *: 0개 이상의 임의 문자열과 매칭
        - \$: URL 끝을 의미
        - 더 구체적인 규칙이 우선한다 ex. allow: /public/ vs disallow / -> /public/은 접근 허용. 나머지는 불허.

            </tr>

### 2. 동적 웹 페이지 크롤링
- 동적 웹 페이지: JavaScript를 이용해 함수를 호출하고 데이터를 처리
    - BeautifulSoup 라이브러리는 정적 웹 페이지에서 크롤링하는 것은 가능하나, 동적 웹 페이지 크롤링에 부적합
    - *해결방법*: Selenium 라이브러리 사용
- Selenium: 웹 애플리케이션 테스트를 위한 포터블 프레임워크

````python
# 최초 1회만 실행
# !pip install selenium

from selenium import webdriver # webdriver 임포트. 자동으로 브라우저 버전과 매핑해줌.
wd = webdriver.Chrome() # chrome webdriver 객체 생성

wd.get("http://www.hanbit.co.kr") # selenium이 제어하는 크롬 창에서 웹페이지를 열어 확인
````
